### Set up TensorFlow

```python
import tensorflow as tf

keras = tf.keras

print("TensorFlow version:", tf.__version__)
```

### Load a Dataset

**MNIST Dataset** : contains pixel values of images, with a range of 0-255. We scale these values to a range of 0-1 by dividing by 255.0, which converts the initial integer values to floats

```python
mnist = keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0
```

### Buils a Machine Learning Model

```python
model = keras.models.Sequential(
    [
        keras.layers.Flatten(input_shape=(28, 28)),
        keras.layers.Dense(128, activation="relu"),
        keras.layers.Dropout(0.2),
        keras.layers.Dense(10),
    ]
)
```

```python
predictions = model(x_train[:1]).numpy()
predictions
```

```python
tf.nn.softmax(predictions).numpy()
```

```python
loss_fn = keras.losses.SparseCategoricalCrossentropy(from_logits=True)
```

```python
loss_fn(y_train[:1], predictions).numpy()
```

```python
model.compile(optimizer="adam", loss=loss_fn, metrics=["accuracy"])
```

```python
model.fit(x_train, y_train, epochs=5)
```

```python
model.evaluate(x_test, y_test, verbose=2)
```

```python
probability_model = keras.Sequential(
    [
        model,
        keras.layers.Softmax(),
    ]
)
```

```python
probability_model(x_test[:5])
```
