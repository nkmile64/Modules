### Create a TextBlob

```python
from textblob import TextBlob
from textblob import Word
from textblob.wordnet import VERB
from textblob.wordnet import Synset

wiki = TextBlob("Python is a high-level, general-purpose programming language.")
```

### Part-of-Speech Tagging

```python
wiki.tags
```

### Noun Phrase Extraction

```python
wiki.noun_phrases
```

### Sentiment Analysis

```python
testimonial = TextBlob("TextBlob is amazingly simple to use. What great fun!")
print(testimonial.sentiment)
print(testimonial.sentiment.polarity)
print(testimonial.sentiment.subjectivity)
```

### Tokenization

```python
zen = TextBlob("Beautiful is better than ugly. " "Explicit is better than implicit. " "Simple is better than complex.")

print(zen.words)
print(zen.sentences)
```

```python
for sentence in zen.sentences:
    print(sentence.sentiment)
```

### Words Inflection & Lemmatization

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
print(sentence.words)
print(sentence.words[2].singularize())
print(sentence.words[-1].pluralize())
```

```python
w = Word("octopi")
print(w.lemmatize())

w = Word("went")
print(w.lemmatize("v"))  # indicate that part-of-speech is verb
```

### WordNet Integration

> **synset (synonym set)** : a group of words that all express the same underlying concept. Each synset represents **one specific meaning** of a word.For example,<br/>

> one meaning : octopus as food<br/>
another meaning : the animal

```python
word = Word("octopus")
word.synsets
```

```python
Word("hack").get_synsets(pos=VERB)
```

```python
Word("octopus").definitions
```

```python
octopus = Synset("octopus.n.02")
shrimp = Synset("shrimp.n.03")
octopus.path_similarity(shrimp)
```

### Get Word and Noun Phrase Frequencies

```python
# 1. Use the word_counts dictionary
monty = TextBlob("We are no longer the Knights who say Ni. " "We are now the Knights who say Ekki ekki ekki PTANG.")
monty.word_counts["ekki"]
```

```python
# 2. Use the count() method
monty.words.count("ekki")
```

```python
monty.words.count("ekki", case_sensitive=True)
```
