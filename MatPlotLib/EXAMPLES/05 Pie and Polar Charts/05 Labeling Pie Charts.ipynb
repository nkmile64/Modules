<h3 style="font-weight:600;color:#A5BA45">1 | With a list of strings</h3>

```python
import matplotlib.pyplot as plt

data = [36, 24, 8, 12]
labels = ["spam", "eggs", "bacon", "sausage"]

fig, ax = plt.subplots()
pie = ax.pie(data)
ax.pie_label(pie, labels)

plt.show()
```

<h3 style="font-weight:600;color:#A5BA45">2 | Place labels outside the pie</h3>

Define a distance from the center of the pie as a fraction of its radius (>1 means label is outside the pie).

```python
fig, ax = plt.subplots()
pie = ax.pie(data)
ax.pie_label(pie, labels, distance=1.2)

plt.show()
```

<h3 style="font-weight:600;color:#A5BA45">3 | Rotate the labels</h3>

```python
fig, ax = plt.subplots()
pie = ax.pie(data)
ax.pie_label(pie, labels, rotate=True)

plt.show()
```

<h3 style="font-weight:600;color:#A5BA45">4 | Label slices with their values</h3>

```python
fig, ax = plt.subplots()
pie = ax.pie(data)
ax.pie_label(pie, "{absval:.1f}")

plt.show()
```

<h3 style="font-weight:600;color:#A5BA45">5 | Label slices with their percentages</h3>

```python
fig, ax = plt.subplots()
pie = ax.pie(data)
ax.pie_label(pie, "{frac:.1%}")

plt.show()
```

<h3 style="font-weight:600;color:#A5BA45">6 | Label slices with their percentages & absolute values</h3>

```python
fig, ax = plt.subplots()
pie = ax.pie(data)
ax.pie_label(pie, "{absval:d}\n{frac:.1%}")

plt.show()
```

<h3 style="font-weight:600;color:#A5BA45">7 | Configure font styling</h3>

Style the font by passing a dict to the **textprops** parameter.

```python
fig, ax = plt.subplots()
pie = ax.pie(data)
ax.pie_label(pie, labels, textprops=dict(fontsize="large", color="yellow"))

plt.show()
```

<h3 style="font-weight:600;color:#A5BA45">8 | Add multiple sets of labels</h3>

Add multiple sets of labels by calling `pie_label` repeatedly.

```python
fig, ax = plt.subplots()
pie = ax.pie(data)

ax.pie_label(pie, labels, distance=1.1)
ax.pie_label(pie, "{frac:.1%}", distance=0.7)
ax.pie_label(pie, "{absval:d}", distance=0.5)

plt.show()
```
