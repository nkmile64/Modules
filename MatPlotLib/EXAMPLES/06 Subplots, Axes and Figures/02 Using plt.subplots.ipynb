```python
import matplotlib.pyplot as plt
import numpy as np

# Some example data to display
x = np.linspace(0, 2 * np.pi, 400)
y = np.sin(x**2)
```

### **A Figure with just One Subplot**

```python
fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("A single plot")
```

### **Stacking Subplots in One Direction**

```python
fig, axs = plt.subplots(2)  # 2 rows, 1 column
fig.suptitle("Vertically stacked subplots")
axs[0].plot(x, y)
axs[1].plot(x, -y)
```

Unpack the Axes immediately to dedicated variables - instead of indexing such as axs[0]

```python
fig, (ax1, ax2) = plt.subplots(2)  # 2 rows, 1 column
fig.suptitle("Vertically stacked subplots")
ax1.plot(x, y)
ax2.plot(x, -y)
```

For **side-by-side** subplots e.g. 1 row 2 columns:

```python
fig, (ax1, ax2) = plt.subplots(1, 2)  # 1 row, 2 columns
fig.suptitle("Horizontally stacked subplots")
ax1.plot(x, y)
ax2.plot(x, -y)
```

### **Stacking Subplots in Two Directions**

```python
fig, axs = plt.subplots(2, 2)  # 2 rows, 2 columns

axs[0, 0].plot(x, y)
axs[0, 0].set_title("Axis [0,0]")
axs[0, 1].plot(x, y, "tab:orange")
axs[0, 1].set_title("Axis [0,1]")
axs[1, 0].plot(x, -y, "tab:green")
axs[1, 0].set_title("Axis [1,0]")
axs[1, 1].plot(x, -y, "tab:red")
axs[1, 1].set_title("Axis [1,1]")

for ax in axs.flat:
    ax.set(xlabel="x-label", ylabel="y-label")

# Hide x labels and tick labels for top plots and y ticks for right plots.
for ax in axs.flat:
    ax.label_outer()
```

We can assign all subplots to dedicated variables

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2)
fig.suptitle("Sharing x per column, y per row")

ax1.plot(x, y)
ax2.plot(x, y**2, "orange")
ax3.plot(x, -y, "green")
ax4.plot(x, -(y**2), "red")

for ax in fig.get_axes():
    ax.label_outer()
```

### **Sharing Axes**

```python
fig, (ax1, ax2) = plt.subplots(2)
fig.suptitle("Axes values are scaled individually by default")

ax1.plot(x, y)
ax2.plot(x + 1, -y)
```

To **align horizontal** or **vertical** axes, use *sharex* or *sharey*.

```python
fig, (ax1, ax2) = plt.subplots(2, sharex=True)
fig.suptitle("Aligning x-axis using sharex")

ax1.plot(x, y)
ax2.plot(x + 1, -y)
```

For vertically stacked subplots, setting `sharey=True` forces the y-axes of these subplots to have the same scale.

```python
fig, axs = plt.subplots(3, sharex=True, sharey=True)
fig.suptitle("Sharing both axes")

axs[0].plot(x, y**2)
axs[1].plot(x, 0.3 * y, "o")
axs[2].plot(x, y, "+")
```

Notice above how use of `sharex=True` has removed the labels for the inner axes.<br/>
If we want to restore them, use `tick_params`.

```python
fig, axs = plt.subplots(3, sharex=True, sharey=True)
fig.suptitle("Restore xtick labels on to Axes")

axs[0].plot(x, y**2)
axs[1].plot(x, 0.3 * y, "o")
axs[2].plot(x, y, "+")

axs[0].tick_params(labelbottom=True)  # Restore xtick labels on to Axes
```

To reduce the height between vertical subplots, use a `GridSpec` and set the `hspace` to the desired value:

```python
fig = plt.figure()
gs = fig.add_gridspec(3, hspace=0)
axs = gs.subplots(sharex=True, sharey=True)
fig.suptitle("Sharing both axes")

axs[0].plot(x, y**2)
axs[1].plot(x, 0.3 * y, "o")
axs[2].plot(x, y, "+")

for ax in axs:
    ax.label_outer()  # Hide x labels and tick labels for top plots and y ticks for right plots.
```

```python
fig = plt.figure()
gs = fig.add_gridspec(2, 2, hspace=0, wspace=0)
(ax1, ax2), (ax3, ax4) = gs.subplots(sharex="col", sharey="row")
fig.suptitle("Sharing x per column, y per row")

ax1.plot(x, y)
ax2.plot(x, y**2, "tab:orange")
ax3.plot(x + 1, -y, "tab:green")
ax4.plot(x + 2, -(y**2), "tab:red")

for ax in fig.get_axes():
    ax.label_outer()
```

### **Polar Axes**

```python
fig, (ax1, ax2) = plt.subplots(1, 2, subplot_kw={"projection": "polar"})

ax1.plot(x, y)
ax2.plot(x, y**2, "darkslateblue")

plt.show()
```
