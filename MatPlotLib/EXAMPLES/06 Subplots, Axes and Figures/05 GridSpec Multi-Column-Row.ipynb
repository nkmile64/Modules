A 3X3 grid and axes spanning all 3 columns, 2 columns and 2 rows.

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from matplotlib.figure import Figure


def format_axes(fig: Figure):
    for i, ax in enumerate(fig.axes):
        ax.text(0.5, 0.5, f"ax{i+1}", va="center", ha="center")
        ax.tick_params(labelbottom=False, labelleft=False)


fig = plt.figure(layout="constrained")
gs = GridSpec(3, 3, figure=fig)

ax1 = fig.add_subplot(gs[0, :])  # first row, all columns
ax2 = fig.add_subplot(gs[1, :-1])  # second row, all but the last column
ax3 = fig.add_subplot(gs[1:, -1])  # second and last row, last column
ax4 = fig.add_subplot(gs[-1, 0])
ax5 = fig.add_subplot(gs[-1, -2])

fig.suptitle("GridSpec")
format_axes(fig)

plt.show()
```
