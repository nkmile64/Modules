```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.axes import Axes


def example_plot(ax: Axes):
    ax.plot([1, 2])
    ax.set_xlabel("x-label", fontsize=12)
    ax.set_ylabel("y-label", fontsize=12)
    ax.set_title("Title", fontsize=14)
```

If we don't use **constrained layout**, labels overlap the Axes

```python
fig, axs = plt.subplots(nrows=2, ncols=2, layout=None)

for ax in axs.flat:
    example_plot(ax)
```

Remove overlap between axes and labels using `layout="constrained"`.

```python
fig, axs = plt.subplots(nrows=2, ncols=2, layout="constrained")

for ax in axs.flat:
    example_plot(ax)
```

A more complicated example using **nested gridspecs**.

```python
fig = plt.figure(layout="constrained")

gs0 = gridspec.GridSpec(1, 2, figure=fig)
gs1 = gridspec.GridSpecFromSubplotSpec(3, 1, subplot_spec=gs0[0])  # first column of gs0
gs2 = gridspec.GridSpecFromSubplotSpec(2, 1, subplot_spec=gs0[1])  # second column of gs0

for n in range(3):
    ax = fig.add_subplot(gs1[n])
    example_plot(ax)

for n in range(2):
    ax = fig.add_subplot(gs2[n])
    example_plot(ax)

plt.show()
```
