**SMOTE (Synthetic Minority Over-sampling Technique)** is used to perform *over-sampling* in cases where we have an *imbalanced* dataset i.e. it generates new  synthetic samples using interpolation, for the classes that are *under-represented*.

<pre style="font-family:'Roboto Mono', 'Cascadia Mono Light', 'Courier New';font-size:0.85em;">
<span style="color:orange;font-weight:600;">imblearn.over_sampling</span>.<span style="color:darkblue;font-weight:600;">SMOTE</span>(
    sampling_strategy,  # Default is "auto" which means resample all classes but the majority class.
                        # After resampling, the number of samples in the different classes will be equal.
    random_state,       # If int, it will be used as the seed for the random number generator.
                        # Default is None.
    k_neighbors,        # Default is 5. This is the nearest neighbors used to define the neighborhood
                        # of samples to use, to generate the synthetic samples.
)
</pre>

```python
from collections import Counter
from sklearn.datasets import make_classification
from imblearn.over_sampling import SMOTE

# Generate a random n-class classification problem.
X, y = make_classification(
    n_classes=2,
    class_sep=2,
    weights=[0.1, 0.9],  # proportions of samples assigned to each class
    n_informative=3,  # number of informative features
    n_redundant=1,  # number of redundant features
    flip_y=0,  # fraction of samples whose class is assigned randomly
    n_features=20,  # total number of features
    n_clusters_per_class=1,
    n_samples=1000,  # number of samples
    random_state=10,
)
print(f"Original dataset shape\t{Counter(y).values()}")

sm = SMOTE(random_state=42)
X_res, y_res = sm.fit_resample(X, y)

print(f"Resampled dataset shape\t{Counter(y_res).values()}")
```
