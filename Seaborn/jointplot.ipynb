### seaborn.JointPlot

`JointPlot` draws a plot of two variables with bivariate and univariate graphs.<br/>

<pre style="font-family:'Courier New';font-size:1.0em;">
<span style="color:orange;font-weight:600;">sns</span>.<span style="color:darkblue;font-weight:600;">JoinPlot</span>(
    data,           # Input data structure. Could be a long-form collection of vectors or
                      a wide-form dataset that will be internally reshaped.
    x, y,           # Variables that specify positions on the x and y axes.
    hue,            # The name of a column to map plot aspects to different colors.
    kind,           # Kind of plot to draw: scatter|kde|hist|hex|reg|resid.
    height,         # Size of the figure (it will be square).
    ratio,          # Ratio of joint axex height to marginal axes height
    space,          # Space between the joint and marginal axes.
    dropna          # If True, remove observations that are missing from x and y.
    {x, y}lim       # Axis limits to set before plotting.
    color           # Single color for when hue mapping is not used.
    palette,        # Set of colors for mapping the `hue` variable.
    hue_order,      # The order for processing and plotting for categorical levels of the hue variable.
    hue_norm,       # A pair of values setting the normalization range or an object 
                      mapping data units to a [0,1] interval.
    marginal_ticks, # If False, suppress ticks on the count/density axis of the marginal plots.
)
</pre>

### Examples

By default it creates a `scatterplot` with marginal histograms:

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
)
plt.show()
```

Assigning a `hue` variable, adds conditional colors to the scatterplot and draws separate density curves on the marginal axes:

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
)
plt.show()
```

Setting `kind="kde"` draws both bivariate and univariate KDEs:

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    kind="kde",
)
plt.show()
```

Set `kind="reg"` to add a linear regression fit and univariate KDE curves:

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    kind="reg",
)
plt.show()
```

For bin-based visualization of the joint distribution we can use `kind="hist"`

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    kind="hist",
)
plt.show()
```

or use `kind="hex"` to compute a bivariate histogram using hexagonal bins:

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    kind="hex",
)
plt.show()
```

Try using additional keyword arguments:

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    marker="+",
    s=100,
    marginal_kws=dict(bins=25, fill=False),
)
plt.show()
```

Control the size and layout of the figure:

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    height=5,
    ratio=2,
    marginal_ticks=True,
)
plt.show()
```

Add more layers onto the plot

```python
g = sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
)
g.plot_joint(
    sns.kdeplot,
    color="r",
    zorder=0,
    levels=6,
)
g.plot_marginals(
    sns.rugplot,
    color="r",
    height=-0.15,
    clip_on=False,
)
plt.show()
```
